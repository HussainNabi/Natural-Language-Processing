```python
from transformers import pipeline, set_seed
from transformers import AutoTokenizer, AutoModelForSeq2SeqLM
```

```python
## Task 2 :: Use the sentiment classifier on these three sentences.

classifier = pipeline("sentiment-analysis", model="distilbert-base-uncased-finetuned-sst-2-english")

print(classifier("The movie was excellent."))
print(classifier("The lecture was difficult to understand."))
print(classifier("I enjoyed learning about transformers."))
```

```python
## Task 3 — Encoder–Decoder Model

tokenizer = AutoTokenizer.from_pretrained("google/flan-t5-small")
model = AutoModelForSeq2SeqLM.from_pretrained("google/flan-t5-small")

prompt = "Summarize the following paragraph: I like to eat ice cream, chocolate, and other sweet treats. I also enjoy going to the park and playing with my friends. My favorite color is blue, and I love to read books about animals."
# Convert text into tokens
output = model.generate(**tokenizer(prompt, return_tensors="pt"),max_new_tokens=60, num_beams=4, no_repeat_ngram_size=3)

# Convert tokens back to text
print(tokenizer.decode(output[0], skip_special_tokens=True))
```

```python

## Task 5 — Short Reflection
###  Why is a decoder-only model suitable for a chatbot such as ChatGPT?
'''A chatbot's job is to continue a conversation, which means generating the next word again and again.
 A decoder-only model is trained for exactly that: it predicts the next token from all the tokens before
   it. The user's message and the earlier conversation are placed in the same sequence as the reply, so 
   there is no need for a separate encoder to read the input. This also makes one model
 and one training objective enough, which scales well to very large models and large amounts of text.'''
```
