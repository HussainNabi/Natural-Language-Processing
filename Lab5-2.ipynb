```python
## Imports
#!pip install gensim
from sklearn.feature_extraction.text import TfidfVectorizer
import pandas as pd
from sklearn.metrics.pairwise import cosine_similarity
import nltk
import numpy as np
import gensim
from nltk.tokenize import word_tokenize
```

```python
### Task 1: Cosine Similarity

data = [ 'This is the first document.',  'This document is the second document.',  
    'And this is the third one.',  'Is this the first document?'      ]

tfidf_vectorizer = TfidfVectorizer()
vector_matrix = tfidf_vectorizer.fit_transform(data)

cosine_similarity_matrix = cosine_similarity(vector_matrix)
df_cosine = pd.DataFrame(data=cosine_similarity_matrix, index=data, columns=data)
df_cosine
```

```python
### Task 2: TF-IDF

data = ['data science is one of the most important fields of science'
        ,  'this is one of the best data science courses'
        ,  'data scientists analyze data'  ]


tfidf = TfidfVectorizer()
result = tfidf.fit_transform(data) # returns sparce matrix

df = pd.DataFrame(result.toarray(), columns=tfidf.get_feature_names_out())
df
```

```python
### Task 3: Word2Vec
import re
from xxlimited import Str

df = pd.read_csv("simpsons_script_lines.csv")

def clean_text(text):
    text = Str(text)
    text = text.lower()
    text = re.sub(r"[0-9]", '', text)
    text = re.sub(r"[)(,”“.’$-]", '', text)
    return text

df['cleaned_text'] = df['spoken_words'].apply(clean_text)

tokens = df['cleaned_text'].str.split().tolist()

Skip_gram_model = gensim.models.Word2Vec(tokens, min_count = 1, vector_size = 100, window = 5, sg = 1)
```

```python
### Task 4 Use: wv.most_similar() method 
print(Skip_gram_model.wv.most_similar('homer', topn=5))
print(Skip_gram_model.wv.most_similar('marge', topn=5))
print(Skip_gram_model.wv.most_similar('bart', topn=5))
```

```python
print(Skip_gram_model.wv.doesnt_match(['jimbo','kearney','milhouse']))
print(Skip_gram_model.wv.doesnt_match(['nelson','bart','milhouse']))
print(Skip_gram_model.wv.doesnt_match(['homer', 'patty', 'selma']))
```
