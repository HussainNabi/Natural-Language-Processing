```python
import re
```

```python
#1- findall
text = "Woodchucks is another name for groundhog!"

re.findall(r"groundhog|woodchuck",text)
#^^ case sensitive, only found "groundhog" because 'W' in 
# "Woodchuck" is Capital in text and Small in call.
```

```python
#2- Search
text = 'I am enjoying the NLP course.'

print(re.search(r"I", text)) #r"I" means it's a raw string literal
print(re.search(r"se.$", text)) #$ Search at the end of the string
print(re.search(r"am", text))
print(re.search(r"m", text))
print(re.search(r"AI", text))
```

```python
#3- Match and its functions

match = re.search((r"am"),text)

print(match.span())
print(match.start())
print(match.end())

print("The matched pattern is: ",text[match.start():match.end()]) 
#^^extracting substring from string " Text[x:y]"
#return matched string using index from text

print("The matched pattern is: ", match.group())
 #return matched string using group() method
```

```python
#5- Search vs Match
text = 'I am enjoying the NLP course.'

print(re.search(r"I.*", text)) #global
#^^search scans the whole string for a match anywhere
print(re.match(r"I.*", text)) #local >> search at the start of the string
#^^match only checks if the pattern matches starting at position 0.

print(re.search(r"enjoying.*", text))
#^^found "enjoying" at index 5, took it and everything after.
print(re.match(r"enjoying.*", text))
#^^did not find "enjoying" at index zero, nothing happens.
```

```python
#6- find / match - Letters

text = '1999 was the year I was born.'

print(re.search(r"[a-zA-Z]+", text))
#^^ checks for the first instant of Letters (Index 5)
print(re.match(r"[a-zA-Z]+", text))
#^^ checks if index 0 is a Letter, it is not.
```

```python
#7- finall - digits
text1 = "a 1 b 2 c 3"
text2 = "a1 b2 c 3"
text3 = "a 1 b 22 2 c 3"

print(re.findall(r"\d+", text1)) #match any digit
print(re.findall(r"\d+", text2))
print(re.findall(r"\d+", text3))
#^^Looks for seperate number instances
```

```python
#9- Substitute
text = 'I am enjoying the Math course.'

print(re.sub(r"Math", "NLP", text, count=1, flags=re.I))
 #flags=re.I >> Ignore the case (no matter if M or m)
 # count = 1 >> number of word I need it to change. count= 0 by default
```

```python
#10- compile?
re.compile?
```

```python
#11- Patterns
text = "This year is 2021"
pattern = re.compile(r"\d+")
#^^ a set pattern to be found and reused in text without the need to specify everytime
print(type(pattern))

print(pattern.sub("2022", text))
```

```python
#19- Spacy tokenization
#!python -m spacy validate
import spacy
nlp = spacy.load('en_core_web_sm') # Load the small English model - - nlp becomes a pipeline you can use to process text (e.g., tokenize, tag, parse).

text = "A 5km NYC cab ride costs $10.30"
doc = nlp(text=text) # Process the text using nlp model, returns Doc object.
 # Process the text using nlp model, returns Doc object.

for token in doc: # Iterate over each token in the Doc object.
    print(token)
```

```python
#20- Spacy - NLTK Tokenization
text = "A 5km NYC cab ride costs $10.30"
doc = nlp(text=text) # Process the text using nlp model, returns Doc object.

print('nltk Tokenizer result: ')
print(nltk.word_tokenize(text))

print('\nSpaCy Tokenizer result: ')

for token in doc:
    print(token)
```

```python
#22- Stemming
from nltk.stem import PorterStemmer

ps = PorterStemmer() # create a PorterStemmer object

words = ['run','runner','running','ran','runs','easily','fairly']

for word in words:
    print(word + ' --> ' + ps.stem(word))
```

```python
#23- Snowball Stemming
from nltk.stem.snowball import SnowballStemmer

sn = SnowballStemmer(language='english')

words = ['run','runner','running','ran','runs','easily','fairly']

for word in words:
    print(word+' --> '+sn.stem(word))
```

```python
#24- Porter vs Snowball
words = ['generous','generation','generously','generate']

for word in words:
    print(word+' --> '+ps.stem(word))
    print(word+' --> '+sn.stem(word))
    print('---------------------------------------')
```

```python
#26- Lemmatization
#returns the base or dictionary form of a word, which is known as the lemma.
# #nltk.download('wordnet')
from nltk.stem import WordNetLemmatizer

text = 'I am enjoying AI courses. I am taking NLP course.'
lemmatizer = WordNetLemmatizer()

for word in nltk.word_tokenize(text):
    print(f"{word}: ", lemmatizer.lemmatize(word))
```

```python
#27- Changing pos to verb
for word in nltk.word_tokenize(text):
    print(f"{word}: ", lemmatizer.lemmatize(word, pos='v'))
```
